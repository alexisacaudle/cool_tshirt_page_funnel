# Cool T-Shirts Page Visits Funnel

```python
import pandas as pd

visits = pd.read_csv('visits.csv',
                     parse_dates=[1])
cart = pd.read_csv('cart.csv',
                   parse_dates=[1])
checkout = pd.read_csv('checkout.csv',
                       parse_dates=[1])
purchase = pd.read_csv('purchase.csv',
                       parse_dates=[1])
visits.head(10)
```

## Compare the number of visits to the number to the number of purchases

```python
# Merge the visits and cart dataframes
visits_cart = pd.merge(visits, cart, how='left')

print(visits_cart.shape[0])

# Total number of users who visited, but did not add an item to the cart
no_cart = visits_cart.cart_time.isnull().sum()

# Total number of visitors
total_visits = visits.user_id.count()

# Calculate the percentage of how many users visited but add an item to the cart
visit_no_purchase = round((no_cart / total_visits)*100, 2)
```

## Compare the number of times an item was placed in the cart to the number of checkouts began.

```python
# Merge the cart and checkout dataframaes
cart_checkout = pd.merge(cart, checkout, how = 'left')

# Total number of users who added an item to the cart
added_to_cart = cart.user_id.count()

# Total number of users who added an item to the cart, but did not checkout
cart_no_checkout = cart_checkout.checkout_time.isnull().sum()

# Calculate the percentage of how many users added items to the cart, but did not checkout
not_checked_out = round((cart_no_checkout / added_to_cart) *100, 2)
```

## Compare the number of checkouts began to the number of purchases made.

```python
# Merge all dataframes into one
all_data = visits.merge(cart, how='left').merge(checkout, how='left').merge(purchase, how='left')

# Filter the users who began the checkout process, but did not make a purchase
began_checkout_no_purchase = all_data[
    (all_data.checkout_time.notnull()) & (all_data.purchase_time.isnull())
]

# Count the number of users who began the checkout process
began_checkout = all_data.checkout_time.notnull().sum()

# Count the number to users who began checkout, but didn't make a purchase
count_began_checkout_no_purchase = len(began_checkout_no_purchase)


# Calculate the percentage of users who began checkout, but didn't make a purchase
percentage_began_checkout_no_purchase = round((count_began_checkout_no_purchase / began_checkout) *100, 2)
```

## Average checkout time

```python
# Calculate average time from initial visit to final purchase
all_data['time_til_purchase'] = all_data.purchase_time - all_data.visit_time

average_checkout_time = all_data.time_til_purchase.mean()
all_data.head(10)
```

## Findings

```python
print(f'{total_visits} total visits\n{no_cart} did not make a purchase\n{visit_no_purchase}% of users visited the website, but did not check out.\n')

print(f'{added_to_cart} users added items to the cart\n{cart_no_checkout} users added items to the cart but did not checkout\n{not_checked_out}% users added items to the cart but did not checkout\n')

print(f'{began_checkout} users began the checkout process.')

print(f'{count_began_checkout_no_purchase} users began the checkout process, but did not make a purchase.')

print(f'{percentage_began_checkout_no_purchase}% of users began the checkout process, but did not make a purchase.\n')

print(f'The weakest part of the funnel is the drop-off from visiting the website to adding items to the cart or checking out, with an {visit_no_purchase}% drop-off.')

print(f'The average time from initial visit to final purchase is {average_checkout_time}')
```
